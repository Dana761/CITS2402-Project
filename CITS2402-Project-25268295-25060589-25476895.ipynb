## **Introduction**

Housing insecurity is one of the most prominent problems both Australian and New Zealand citizens are facing today. Whether driven by insecure tenure, unaffordable costs or inadequate living space, housing insecurity is a concern that continues to grow, shaping policy debates around housing supply and rental protections. Although national housing statistics are publicly reported, very little attention is given to how insecurity is specifically distributed amongst the population - specifically, whether certain household compositions or dwelling types are more likely to be susceptible to housing insecurity than others. Single national affordability figures available to the public are often abstracted, concealing the concentration of risks in specific groups, meaning population-level statistics alone are insufficient for identifying who is actually most at risk.  

This report aims to address this gap and investigate housing security through 3 central domains - housing tenure, housing affordability and housing adequacy.

The intuition is that each domain captures a distinct form of insecurity:
- Tenure reflects contractual and long-term stability

- Affordability reflects the financial strain of maintaining that housing

- Adequacy reflects whether the housing itself physically meets a household's needs.

The analysis of these 3 domains illustrate how a household can be secure on one measure while exposed on another. For example, a family may be well-housed physically yet insecure in tenure. Examining all three together, rather than in isolation, is therefore necessary to identify where housing insecurity is most prevalent across Australia and New Zealand.

## **Additional Data Information and Limitations in Comparisons**





* G37 (Tenure). AU's tenure categories don't map one-to-one onto NZ's tenure data (mortgage/rent payments made or not made, family-trust arrangements) — the two systems classify tenure on different underlying logic, so categories are harmonised only where concepts are close enough, not matched exactly.


* G33 / G40 (Affordability). AU splits affordability data into two disconnected tables — income by household composition (G33) and rent by landlord type (G40) — with no shared category to join them on. This means rent-based affordability can only be assessed at the aggregate renter population level in AU.  NZ's data  allows for a more granular household-level cost-to-income comparison but AU's disconnected data limits how precisely the two countries can be compared on rental affordability specifically.


* G35 (Household composition). AU only publishes a single combined Family vs Non-family total, with no further breakdown by relatedness or number of families — a household counts as Family if it contains at least one family unit (two or more related people), regardless of how many families or unrelated individuals are also present. NZ's data is more granular (distinguishing related, unrelated, and mixed multi-person households, and one-family vs multi-family households) however due to how AU's data is structured would have to be simplified into simply Family and Non-Family categories for comparison. Additionally, the number of usual non-related residents columns for AU stops at 6 and above while NZ stops at 8 and above. As such we'll have to merge the NZ 6,7 and 8+ residents columns into one.


* G41 / G42 / G29 (Adequacy). NZ's data contains adequate housing measures such as number of residents compared to number of bedrooms. However,   AU has no table linking bedroom count to household size or number of occupants — G41 (bedrooms) and G35 (household size) cannot be validly joined, and G42 (which links dwelling structure to household/family composition) does not include bedroom count, so it can't bridge the two either. Therefore there is no reliable way to compare bedroom to resident ratio between NZ and AU.





## **Data Assumptions**

- Different Census years are treated as broadly comparable. Australian tables such as G33, G37, G40, G41, and G42 use 2021 Census data, while the equivalent New Zealand datasets use 2023 Census data. The analysis assumes the two-year gap still allows comparison of broad housing patterns, while recognising that some housing conditions may have changed over that period.

- Australian and New Zealand categories are treated as conceptually comparable, not identical. For example, G37 – Tenure and Landlord Type by Dwelling Structure uses Australian tenure categories, while the NZ “Totals by topic for households” table uses categories such as mortgage payments made/not made, rent payments made/not made, and family-trust arrangements. These are harmonised only where the underlying concepts are sufficiently similar.

- Aggregate totals and subtotals are excluded when detailed categories are used. Tables such as G33 – Total Household Income (Weekly) by Household Composition and G41 – Dwelling Structure by Number of Bedrooms contain detailed categories alongside totals. Keeping both would cause double-counting, so totals are removed before analysis.

- Household and dwelling counts are assumed to be the appropriate unit of analysis. Tables such as G33, G37, G40, G41, and G42 describe housing at the household or occupied-dwelling level, which is appropriate for analysing tenure, affordability, and housing adequacy rather than individual-level population characteristics.

- “Not stated”, “not elsewhere included”, and similar categories are not automatically reassigned. For example, the NZ tenure table includes “Not elsewhere included”, while Australian tables contain categories such as dwelling structure not stated or income not stated. These are treated separately unless there is a clear justification for grouping them with another category.

#### **Cleaning Helper Functions**

To clean the data we built shared helper functions to promote reusability across tables.

- `drop_total_columns_au`: removes aggregate total columns from AU's wide-format tables, where totals appear as their own columns
- `drop_total_rows_nz`: removes aggregate total rows from NZ's long-format tables, where totals instead appear as a value within a category column (only removed where genuine detail categories exist alongside the total in the same column).
- `convert_bracket_to_aud`: converts NZ columns dealing with cost (rent, income) in NZD to their AUD equivalent using a fixed exchange rate, so dollar-based comparisons between AU and NZ are on a common currency. This was roughly 1 NZD = 0.92 AUD in 2023 [14]
- `sort_by_bracket`: sorts a table by the numeric lower bound of its bracket labels, since NZ's cost tables aren't guaranteed to come out in bracket order from Aotearoa Data Explorer

Because AU and NZ encode their totals differently (as columns versus as row values) due to being structured as wide and long tables respectively, two separate functions were needed rather than one shared function.

```python
def drop_total_columns_au(df):
    """
    Removes aggregate total columns from Australian Census table to prevent double-counting.
    """

    total_pattern = re.compile(r"(?:^|_)(?:Tot|Total)(?:_|$)", flags=re.IGNORECASE) # Matches 'Tot' or 'Total' as a separate part of a column name
    total_columns = [col for col in df.columns if total_pattern.search(str(col))]  # collect every column name matching the Tot/Total pattern
    return df.drop(columns=total_columns)     # Returns the DataFrame after removing all detected total columns


def drop_total_rows_nz(df):
    """
    Removes aggregate 'Total - ...' rows from NZ Census tables when the same column also contains detailed categories.

    Keeps geographic totals such as:
    'Total - New Zealand by regional council'
    """

    df = df.copy() # Creates a copy so the original DataFrame is not modified directly
    rows_to_drop = pd.Series(False, index=df.index ) # Initially marks every row as "do not drop" and uses the same row indexes as the DataFrame
    excluded_columns = {"Area", "Census year", "OBS_VALUE" } # Excludes Area because NZ geographic totals should normally stay

    for col in df.columns:
        if col in excluded_columns:
            continue # Skips columns that should not be checked for total rows

        values = (df[col].astype("string").str.strip())
        total_rows = values.str.startswith("Total -", na=False)

        detail_rows = (values.notna() & ~total_rows) # Checks that the value is not missing and that the value is not a "Total - " value

        if total_rows.any() and detail_rows.any(): # Only remove totals if the same column contains BOTH: total categories and detailed categories
            rows_to_drop |= total_rows # Adds the detected total rows to the overall list of rows to remove

    return (df.loc[~rows_to_drop].reset_index(drop=True))


def convert_bracket_to_aud(bracket, rate=0.92): # assuming 1 NZD = 0.92 AUD in 2023
    """
    Takes the input as a cost bracket and returns a converted bracket from NZD to AUD
    We only need to deal with costs in Q2 and there are only 5 possible brackets which we must account for
    1. $A-$B
    2. $A or less
    3. $A or more
    4. $A and over
    5. Under $A

    """
    bracket = bracket.replace('$', '') # remove the $ sign
    bracket = bracket.replace(',', '') # remove the ,
    if '-' in bracket:
        vals = bracket.split('-') # split at - to get individual values
        new_val0 = round(int(vals[0].strip()) * rate) # multiply by rate to obtain converted value and round to return an integer
        new_val1 = round(int(vals[1].strip()) * rate) # .strip() is needed as there may be whitespaces from splitting
        return f"${new_val0:,}-${new_val1:,}" # return the new formatted AUD range converted from NZD
    elif 'or' in bracket:
        vals = bracket.split(' or') # split at or to get individual values
        new_val = round(int(vals[0]) * rate)
        if 'less' in bracket:
            return f"${new_val:,} or less"
        elif 'more' in bracket:
            return f"${new_val:,} or more"
    elif 'Under' in bracket:
        vals = bracket.split(' ') # split at ' ' to get individual values
        new_val = round(int(vals[1]) * rate)
        return f"Under ${new_val:,}"
    elif 'and' in bracket:
        vals = bracket.split(' and ') # split at and to get individual values
        new_val = round(int(vals[0]) * rate)
        return f"${new_val:,} and over"

def sort_by_bracket(df, bracket_col):
    """
    Sorts a DataFrame by the numeric lower bound extracted from a bracket column
    ('Under $X' sorts as 0), since these NZ tables aren't guaranteed to come out in order.
    """
    df = df.copy()
    df['_sort_key'] = df[bracket_col].apply(lambda b: 0 if 'Under' in b else int(re.search(r'\d+', b.replace(',', '')).group()))
    return df.sort_values('_sort_key').drop(columns='_sort_key').reset_index(drop=True)
```

#### **Individual Table Processing & Cleaning**

The inherent differences in data formatting between AU and NZ tables means that we have slightly different approaches to cleaning tables form the 2 countries.
- AU's tables are wide-format with column names that encode multiple variables together in inconsistent codes (e.g. `O_MTG_DS_Sep_house`) and as a result individual parse functions are required for every table to decode the column names and convert the tables to a long format

- NZ's tables are already in long format with clean, human-readable category labels, so a single shared function, `clean_nz`, handles every NZ table by simply selecting the relevant columns and removing aggregate total rows — no table-specific decoding is needed (with the exception of Question 4's NZ table where we created a parse function to extract numeric values from the descriptions for future analysis)

`clean_nz` is the shared function used across NZ's tenure, income, and rent tables. It drops aggregate total rows, selects the columns relevant to that specific table, renames the value column to something readable, and standardises column name casing.

```python
def clean_nz(df, columns, value_col_name):
    """
    Cleans an NZ SDMX-format table: drops aggregate total rows, selects the
    requested columns plus the value column, and standardises naming.
    """
    df = drop_total_rows_nz(df) # drop total rows
    columns = columns + ['OBS_VALUE'] # create a list of the cols wanted + OBS_VALUE (value measured)
    df = df[columns] # select the columns
    df = df.rename(columns={'OBS_VALUE': value_col_name}) # rename to a readable column name
    df.columns = df.columns.str.title() # Normalise to make the columns titled
    return df
```

#### **Q1 Table Cleaning**

Two tables are cleaned for this question.
- AU's `G37` (tenure by dwelling structure) is in a wide-format and encodes tenure type and dwelling structure together in each column name (e.g. `O_MTG_DS_Sep_house`). We therefore created a dedicated parse function to decode the description. Only tenure type is retained in the output as dwelling structure is not needed for this question

- NZ's `nzq1` (tenure type and household count) is already long-format with readable category labels, so we only need to apply the `clean_nz` function. Its `"Not elsewhere included"` category is also dropped, as it doesn't correspond to any genuine tenure type.

```python
def parse_G37(df):
    """
    AU G37 — Tenure and Landlord Type by Dwelling Structure, national total.
    Wide format, decodes tenure type from each column name.

    Output: clean table.
    Columns: Tenure Type, Household Count
    """
    df = drop_total_columns_au(df.copy()).drop(columns='AUS_CODE_2021', errors='ignore')
    df = df.melt(var_name='raw_col', value_name='Household Count')

    # convert the encoded tenure description to a readable description
    tenure_patterns = {
        r'^O_OR_': 'Owned outright',
        r'^O_MTG_': 'Owned with mortgage',
        r'^R_': 'Rented',
        r'^Oth_ten_(?:type|ty)_': 'Other tenure'}

    df['Tenure Type'] = None
    # match each raw column name against the tenure patterns and assign its readable label
    for pattern, label in tenure_patterns.items():
        df.loc[df['raw_col'].str.contains(pattern, regex=True, na=False), 'Tenure Type'] = label

    df = df.dropna(subset=['Tenure Type']).copy() # drop rows that didn't match any tenure pattern (e.g. unclassified/not stated columns)
    df['Household Count'] = pd.to_numeric(df['Household Count'], errors='coerce') # convert household counts to numeric, coercing any invalid values to NaN

    return df.groupby('Tenure Type', as_index=False)['Household Count'].sum()
```

#### **Q2 Table Cleaning**

- NZ's tables (`household_income_rent`, `income_dist`, `rent_dist`) are already long-format with clean column names, so we just need to apply the shared `clean_nz` function. We however need to apply a currency conversion function to standardise income and rent costs to AUD across both tables as well as sort them afterwards to prepare for analysis.

```python
def parse_G33(df):
    """
    Parses Australian Census G33:
    Total Household Income (weekly) by Household Composition.

    Output columns:
    - Income Bracket
    - Household Count
    """

    df = drop_total_columns_au(df.copy()).drop(columns="AUS_CODE_2021", errors="ignore") # Remove aggregate totals and geographic identifier
    df = df.melt(var_name="raw_col", value_name="Household Count") # Convert the wide Census table into long format

    # Identify income bracket
    def decode_income(col):
        match = re.match(r"HI_(\d+)_(\d+|more)_", col)

        if match:
            lower, upper = int(match.group(1)), match.group(2)
            if upper == "more":
                return f"${lower:,} or more"
            return f"${lower:,}-${int(upper):,}"
        return None

    df["Income Bracket"] = df["raw_col"].apply(decode_income) # Apply income bracket decoding
    df = df.dropna(subset=["Income Bracket"]).copy() # Remove rows that could not be classified
    df["Household Count"] = pd.to_numeric(df["Household Count"], errors="coerce") # Convert household counts to numeric

    # Sum across all household composition types now that the split isn't needed
    df = df.groupby("Income Bracket", as_index=False)["Household Count"].sum()

    # Sort brackets numerically by their lower bound, since groupby leaves them in arbitrary order
    df["sort_key"] = df["Income Bracket"].apply(lambda b: int(re.search(r"\d+", b.replace(",", "")).group()))
    df = df.sort_values("sort_key").drop(columns="sort_key").reset_index(drop=True)

    return df

def parse_G40(df):
    """
    AU G40 — Rent (Weekly) by Landlord Type, national total.
    Wide format, 1 row, 112 data columns encoding rent_bracket x landlord_type.

    Landlord type is collapsed here (summed across) since it is not relevant to this research question 

    Output: clean, long-format table.
    Columns: rent_bracket, num_households
    """

    df = df.drop(columns='AUS_CODE_2021') # drop the unnecessary aus code
    df = drop_total_columns_au(df) # drop the total columns for the raw non-aggregated table
    df = df.drop(columns=df.columns[df.columns.str.contains('Rent_ns')]) # drop the columns which do not state rent (Rent_ns)
    df = df.melt(var_name='raw_col', value_name='value') # we want to melt the table to become long for easier analysis

    df['raw_col'] = df['raw_col'].str.split('_').str[1:3].str.join("-").str.replace('ovr', 'over', regex=False)
    # extract the integer range from each of the values in raw_col and set that as the new column value and then join to form a string
    # we want to reconstruct the exact excel table which gives the range of rental payments
    # deal with the single ovr typo and replace it with over

    # since we do not need landlord type and would have aggregated along every column anyway (collapse dim 1) we can aggregate by
    # the rental cost range - this will give a table with rental ranges and the number of households that fall within that rental range
    df = df.groupby('raw_col')['value'].sum().reset_index()
    df['sort_key'] = df['raw_col'].str.extract(r'^(\d+)').astype(int) # extract the first number for numeric sorting
    df = df.sort_values('sort_key').drop(columns='sort_key').reset_index(drop=True) # drop the number - only needed for sorting
    df = df.rename(columns={'raw_col': 'Rent Bracket', 'value': 'Number Of Households'}) # rename to reflect final output

    return df
```

```python
# Apply the cleaning to AU and NZ tables
clean_g40 = parse_G40(g40)
clean_g33 = parse_G33(g33)

clean_household_income_rent = clean_nz(household_income_rent, ['Total household income', 'Weekly rent paid by household'], 'Household Count')
clean_income_dist = clean_nz(income_dist, ['Total household income'], 'Household Count')
clean_rent_dist = clean_nz(rent_dist, ['Weekly rent paid by household'], 'Household Count')

# Apply currency conversions via the convert_bracket_to_aud function using .apply() onto columns
clean_household_income_rent['Total Household Income'] = clean_household_income_rent['Total Household Income'].apply(convert_bracket_to_aud, rate=0.92/52) # apply an extra /52 to convert from yearly income to weekly income
clean_household_income_rent['Weekly Rent Paid By Household'] = clean_household_income_rent['Weekly Rent Paid By Household'].apply(convert_bracket_to_aud)
clean_income_dist['Total Household Income'] = clean_income_dist['Total Household Income'].apply(convert_bracket_to_aud, rate=0.92/52) # apply an extra /52 to convert from yearly income to weekly income
clean_rent_dist['Weekly Rent Paid By Household'] = clean_rent_dist['Weekly Rent Paid By Household'].apply(convert_bracket_to_aud)

# Sort brackets numerically now that they're in their final AUD values
clean_income_dist = sort_by_bracket(clean_income_dist, 'Total Household Income')
clean_rent_dist = sort_by_bracket(clean_rent_dist, 'Weekly Rent Paid By Household')
```

#### **Q3 Table Cleaning**

- For NZ table and AU's `G35`, make 2 separate tables each:
-  The first table contains a total family household and non-family household count, with single resident households being grouped with family households, as they are not in a household pooling arrangement. The rest of the columns can be categorised through containing `NonFamHhold`, `FamHhold`, `Household of unrelated people` in different positions in their labels
- The second contains number of resident counts for all non-family categories. As `G35`'s resident count only goes up to 6+, while NZ's goes up to 8+, all NZ household resident counts 6 and above are combined into a single 6+ category. 6+
category is labelled as "6" rather than "6+" to keep the column data type as int for easier comparison later.


```python
def parse_g35_family_nonfamily(df):
    """
    AU G35 — Household Composition by Number of Usual Residents, national total.
    - 'Family Household' — all family households, PLUS one-person households
      (living alone is not a pooling arrangement, so singles are grouped
      with family households rather than with group households)
    - 'Group Household' — non-family households of 2+ unrelated people
    Output columns: Composition Type, Number Of Households
    """
    # pick out every per-size FamHhold/NonFamHhold column, excluding the pre-summed _Total columns
    df = drop_total_columns_au(df)
    value_cols = [c for c in df.columns if c.startswith('Num_Psns_UR_')]

    # reshape from one row of many columns into one row per (size, household type) combination
    long_df = df[value_cols].melt(var_name='raw_col', value_name='Number Of Households')

    # drop the not-applicable cell (1-person FamHhold doesn't exist, marked '..')
    long_df = long_df[long_df['Number Of Households'] != '..'].copy()
    long_df['Number Of Households'] = long_df['Number Of Households'].astype(int)

    # pull household type (Fam/NonFam) and usual residents count out of the original column name
    long_df['Household Type'] = long_df['raw_col'].apply(
        lambda x: 'NonFamHhold' if x.endswith('NonFamHhold') else 'FamHhold')
    long_df['Usual Residents'] = long_df['raw_col'].str.extract(r'UR_(\d)')[0].astype(int)
    long_df.loc[long_df['raw_col'].str.contains('6mo'), 'Usual Residents'] = 6  # open-ended "6 or more" bracket

    def recode(row):
        if row['Household Type'] == 'FamHhold':
            return 'Family Household'
        # one-person NonFamHhold counts as Family (living alone isn't pooling); everything else is Group
        return 'Family Household' if row['Usual Residents'] == 1 else 'Group Household'

    # tag each row with its Composition Type, then collapse into two totals
    long_df['Composition Type'] = long_df.apply(recode, axis=1)
    return long_df.groupby('Composition Type')['Number Of Households'].sum().reset_index()


def parse_g35_group_sizes(df):
    """
    AU G35 — Size distribution of Group Households (2+ unrelated people
    sharing a home), for the pooling-size comparison against NZ.
    Output columns: Usual Residents, Number Of Households
    """
    # only NonFamHhold columns are relevant here (Family households have no size breakdown of interest)
    nonfam_cols = [c for c in df.columns if c.startswith('Num_Psns_UR_') and c.endswith('_NonFamHhold')]
    long_df = df[nonfam_cols].melt(var_name='raw_col', value_name='Number Of Households')
    long_df['Number Of Households'] = long_df['Number Of Households'].astype(int)

    # extract the size from the column name, same convention as above
    long_df['Usual Residents'] = long_df['raw_col'].str.extract(r'UR_(\d)')[0].astype(int)
    long_df.loc[long_df['raw_col'].str.contains('6mo'), 'Usual Residents'] = 6

    # drop one-person households — not a pooling arrangement, excluded from this comparison
    long_df = long_df[long_df['Usual Residents'] > 1]

    # order rows smallest to largest, with clean row numbering
    return long_df[['Usual Residents', 'Number Of Households']].sort_values('Usual Residents').reset_index(drop=True)


def parse_nzq3_family_nonfamily(df):
    """
    NZ Q3 — Household Composition by Number of Usual Residents (no bedroom
    breakdown needed here).
    Recodes NZ's household composition categories into the same two
    categories used for AU:
    - 'Family Household' — one-family / two-family / three-or-more-family
      households, 'household of related people', and 'household of related
      and unrelated people' (a household counts as Family if it contains at
      least one family, per the ABS/Stats NZ definition — even with extra
      unrelated people attached), PLUS one-person households (living alone
      is not pooling)
    Output columns: Composition Type, Number Of Households
    """
    # strip metadata/code columns, drop "Total - ..." rows, rename value column
    df = clean_nz(df, ['Household composition by number of usual residents'], 'Number Of Households')

    # clean_nz title-cases every column name, so store the renamed label column here
    label_col = 'Household Composition By Number Of Usual Residents'

    def recode(label):
        # matched with startswith (not "in") so "household of related and
        # unrelated people" is not wrongly caught by "unrelated"
        if label.strip().lower().startswith('household of unrelated people'):
            return 'Group Household'
        return 'Family Household'  # catches every other category: family types, related, related+unrelated, one-person

    # run recode() once per row, tagging each with its Composition Type
    df['Composition Type'] = df[label_col].apply(recode)

    # collapse every tagged row down into just two totals (one per Composition Type)
    return df.groupby('Composition Type')['Number Of Households'].sum().reset_index()


def parse_nzq3_group_sizes(df):
    """
    NZ Q3 — Size distribution of 'Household of unrelated people' (Group
    Households), for the pooling-size comparison against AU.
    Household sizes 6, 7, and 8-or-more are collapsed into a single
    "6 or more" bucket
    Output columns: Usual Residents, Number Of Households
    """
    df = clean_nz(df, ['Household composition by number of usual residents'], 'Number Of Households')
    label_col = 'Household Composition By Number Of Usual Residents'
    df = df[df[label_col].str.strip().str.lower().str.startswith('household of unrelated people')].copy()

    number_decode = {'one': 1, 'two': 2, 'three': 3, 'four': 4, 'five': 5,
                      'six': 6, 'seven': 7, 'eight': 9}
    count_pattern = r'(one|two|three|four|five|six|seven|eight)'
    df['Usual Residents'] = df[label_col].str.extract(count_pattern)[0].map(number_decode)

    df = df[['Usual Residents', 'Number Of Households']].dropna()

    # merge 6, 7, and "eight or more" (9) into one "6 or more" bucket, matching AU's bracket
    df.loc[df['Usual Residents'] >= 6, 'Usual Residents'] = 6
    df = df.groupby('Usual Residents')['Number Of Households'].sum().reset_index()

    return df.sort_values('Usual Residents').reset_index(drop=True)
```

```python
# Apply the cleaning to AU and NZ tables
nz_family_nonfamily = parse_nzq3_family_nonfamily(nzq3)
nz_nonfamily_composition = parse_nzq3_group_sizes(nzq3)

au_family_nonfamily = parse_g35_family_nonfamily(g35)
au_nonfamily_composition = parse_g35_group_sizes(g35)
```

#### **Q4 Table Cleaning**

- No single AU table links bedroom count directly to household composition, so `G41`, `G42`, `G29`, and `G35` are each cleaned for analysis. `G41` and `G42` are wide tables that use different dwelling-structure codes, so each needs its own parse function. `G41`'s nine raw dwelling categories are collapsed into four (separate house, semi-detached/terrace/townhouse, flat or apartment, other) to match `G42`, and its "not stated" columns are dropped since they can't be aligned across the two tables. `G29` uses a separate `F`/`P` suffix convention to distinguish family counts from person counts, so it also requires a dedicated function to decode the descriptions. `G35` is cleaned using the exact same function (`parse_G35`) as in Q3, given the same Non-family/Lone Person vs Group Household split is required.

- NZ's `nzq4` table already links bedrooms and composition directly in one joint table. However, unlike NZ's other tables it needs a dedicated parse function (`parse_nzq4`) rather than just `clean_nz`. The composition and resident count are combined into a single string label that must be decoded to separate count and composition. The table is then pivoted so bedroom brackets become columns to match AU's derived wide format for comparison

```python
def parse_G29(df):
    """
    AU G29 — Family Composition (families and people, national total).
    Only the 4 aggregate family-type totals are extracted - finer subsplits are not needed

    Output: clean table, one row per family type.
    Columns: Composition Type, Number Of Families, Number Of People
    """

    cols = ['CF_no_children_F', 'CF_no_children_P', 'CF_Total_F', 'CF_Total_P',
            'OPF_Total_F', 'OPF_Total_P', 'Other_family_F', 'Other_family_P']
    df = df[cols].melt(var_name='raw_col', value_name='value')

    # CF_Total was verified against its own sub-total columns: it already represents "couple with children" only (does NOT include CF_no_children),
    family_type_to_clean = pd.Series({
        'CF_no_children': 'Couple, No Children',
        'CF_Total': 'Couple with Children',
        'OPF_Total': 'One Parent Family',
        'Other_family': 'Other Family'})

    # Extract the family type and the indicator if it is a count of families (F) or people (P)
    df['Composition Type'] = pd.Series(df['raw_col'].str.extract(r'(CF_no_children|CF_Total|OPF_Total|Other_family)').values.flatten()).map(family_type_to_clean)
    df['Metric'] = df['raw_col'].str.extract(r'_(F|P)$')[0]

    # Pivot the table so Number of Households and Number of People are columns for every Family Type
    df = df.pivot(index='Composition Type', columns='Metric', values='value').reset_index()
    df = df.rename(columns={'F': 'Number Of Households', 'P': 'Number Of People'})
    df.columns.name = None # drop the name for columns
    return df


def parse_G35(df):
    """
    AU G35 — Household Composition by Number of Usual Residents, national total.
    Only the NonFamHhold columns are extracted — G29 already provides average family size for all family household types

    Output: clean table, one row per household type.
    Columns: Composition Type, Number Of Households, Number Of People
    """

    nonfam_cols = [c for c in df.columns if c.startswith('Num_Psns_UR_') and c.endswith('_NonFamHhold')]
    df = df[nonfam_cols].melt(var_name='raw_col', value_name='Number Of Households')

    # extract the usual-residents digit, resolving the open-ended '6mo' (6 or more) bracket to an assumed value of 7 
    # (same reasoning/value as G41's '6 or more bedrooms' bracket)
    df['Usual Residents'] = df['raw_col'].str.extract(r'UR_(\d)')[0].astype(int)
    df.loc[df['raw_col'].str.contains('6mo'), 'Usual Residents'] = 7

    # lone-person households are trivially 1 person each (Number of People = Number of Households);
    # for group households, total people = residents x household count
    df['Number Of People'] = df['Number Of Households'] * df['Usual Residents']

    # Group into Lone Person and Group Households (to match NZ) and then groupby to get sums
    df['Composition Type'] = df['Usual Residents'].apply(lambda x: 'Lone Person Household' if x == 1 else 'Group Household')
    df = df.groupby('Composition Type')[['Number Of Households', 'Number Of People']].sum().reset_index()

    return df


def parse_G41(df):
    """
    AU G41 — Dwelling Structure by Number of Bedrooms, national total.
    Wide format, 1 row, 63 data columns after cleaning, encoding dwelling_type x bedroom_count.
    9 raw structure categories are collapsed to 4 to align with G42.

    Output: clean, long-format table.
    Columns: Dwelling Type, Bedroom Count, Household Count
    """
    df = df.drop(columns='AUS_CODE_2021') # drop the unnecessary aus code
    df = drop_total_columns_au(df) # drop the total columns for the raw non-aggregated table
    df = df.drop(columns=df.columns[df.columns.str.contains('NS')]) # drop the columns which contain NS
    df = df.melt(var_name='raw_col', value_name='value') # melt to a long table

    # create a series from a dict to map the raw column name codes to a readable dwelling type
    code_to_clean = pd.Series({
        "Separate": 'Separate House',
        "Se": 'Semi-Detached Housing (terrace/townhouse)',
        "Flt": 'Flat or Apartment',
        "Other": "Other Dwelling"})

    df['Dwelling Type'] = df['raw_col'].str.split("_").str[0].map(code_to_clean) # extract dwelling type
    df['Bedroom Count'] = df['raw_col'].str.extract(r'Nof?B_(\d)').astype(int) # extract bedroom count and convert to integer
    df['Bedroom Count'] = df['Bedroom Count'].apply(lambda x: x + 1 if x == 6 else x) # here we assume that 6+ can just be represented as 7
    df = df[['Dwelling Type', 'Bedroom Count', 'value']].rename(columns={'value': 'Household Count'}) # rename and reorder columns
    return df


def parse_G42(df):
    """
    AU G42 — Dwelling Structure by Household and Family Composition, national total.
    Wide format, decodes two dimensions: dwelling structure (same 4-category
    collapse as G41) and household/family composition (6 categories).

    Output: clean, long-format table.
    Columns: Dwelling Type, Composition Type, Household Count
    """

    df = df.drop(columns='AUS_CODE_2021') # drop the unnecessary aus code
    df = drop_total_columns_au(df) # drop the total columns for the raw non-aggregated table
    df = df.drop(columns=df.columns[df.columns.str.contains('NS')]) # drop the columns which contain NS
    df = df.melt(var_name='raw_col', value_name='value')

    # create series' to map the codes to readable column names
    code_to_clean = pd.Series({
        "SH": 'Separate House',
        "SDRTHT": 'Semi-Detached Housing (terrace/townhouse)',
        "FoA": 'Flat or Apartment',
        "OD": "Other Dwelling"})
    composition_to_clean = pd.Series({
        "CF_no_C": "Couple, No Children", # note: Family is removed from CF_no_C and CF_C in order to match with NZ's data
        "CF_C": "Couple with Children",
        "OnePF": "One Parent Family",
        "Oth_fam": "Other Family",
        "Lone_P_H": "Lone Person Household",
        "Group_H": "Group Household"})

    df['Dwelling Type'] = df['raw_col'].str.split("_").str[0].map(code_to_clean) # extract dwelling type
    pattern = r'(CF_no_C|CF_C|OnePF|Oth_fam|Lone_P_H|Group_H)$' # extract either one of these from the text
    # extract code and map to a readable string
    df['Composition Type'] = pd.Series(df['raw_col'].str.extract(pattern).values.flatten()).map(composition_to_clean) 
    df = df[['Dwelling Type', 'Composition Type', 'value']].rename(columns={'value': 'Household Count'})
    return df


def parse_nzq4(df):
    """
    NZ Q4 — Household Composition by Number of Usual Residents and Bedrooms.
    Decodes the combined composition+resident-count text label into two
    separate fields. "Eight or more residents" assumed to be 9.

    Output: wide-format table, one row per composition+resident-count combo.
    Columns: Composition Type, Number Of People, [bedroom bracket columns]
    """
    df = clean_nz(df, ['Household composition by number of usual residents', 'Number of bedrooms'], 'Household Count') # apply the basic NZ cleaning
    # pivot so columns become number of bedrooms
    df = df.pivot(index='Household Composition By Number Of Usual Residents', columns='Number Of Bedrooms', values='Household Count').reset_index() 
    df.columns.name = None

    # decode composition to a standardised composition type and decode text representations of numbers to integers
    composition_decode = pd.Series({
        'Couple with': 'Couple with Children',
        'Couple only': 'Couple, No Children',
        'One parent': 'One Parent Family',
        'One-person': 'Lone Person Household',
        'unrelated': 'Group Household'}) # household of unrelated people and group household are by definition the same across NZ and AU data
    number_decode = {'one': 1, 'two': 2, 'three': 3, 'four': 4, 'five': 5, 'six': 6, 'seven': 7, 'eight': 9} 
    # we assume eight or more can be represented as 9 - same logic as all previous conversions

    # establish the comp and count patterns which we want to extract to be able to map to the standardised composition/integer value for count
    comp_pattern = r'(Couple with|Couple only|One parent|One-person|unrelated)'
    count_pattern = r'(one|two|three|four|five|six|seven|eight)'
    # map to standardised composition type and to integer - split the original household composition title into composition type and household count
    df['Composition Type'] = pd.Series(df['Household Composition By Number Of Usual Residents'].str.extract(comp_pattern).to_numpy().flatten()).map(composition_decode)
    df['Number Of People'] = pd.Series(df['Household Composition By Number Of Usual Residents'].str.extract(count_pattern).to_numpy().flatten()).map(number_decode)

    return df[['Composition Type', 'Number Of People', 'One bedroom', 'Two bedrooms', 'Three bedrooms', 'Four bedrooms', 'Five or more bedrooms']]
```

#### **Sub-Question 1 - Distribution of Household Tenure Across Australia and New Zealand**

How does housing tenure distribution differ between Australia and New Zealand, and which tenure arrangements are associated with greater housing security in each country?

**Harmonising New Zealand Tenure Categories**

Before comparing housing tenure between Australia and New Zealand, the New Zealand tenure categories are harmonised with the broader Australian categories. This is necessary because the two Census systems classify tenure differently, so comparing the original categories directly would not be meaningful.

The harmonisation follows Stats NZ’s own tenure methodology rather than assumptions made by our group. Stats NZ states that owner-occupied housing includes both dwellings that are owned or partly owned and dwellings held in a family trust. Therefore, NZ households in either of these groups are classified according to whether mortgage payments are made: households with no mortgage payments are grouped as Owned outright, while households making mortgage payments are grouped as Owned with mortgage. Households that do not own or hold their dwelling in a family trust and make rent payments are classified as Rented.

Categories that do not have a clear equivalent in the Australian data, such as undefined mortgage or rental arrangements and rent-free arrangements, are excluded from the direct comparison. This harmonisation is completed before calculating proportions, creating graphs, and interpreting the results so that the Australia–New Zealand comparison is based on equivalent tenure concepts rather than differences in Census classification.

```python
def harmonise_nz_tenure(df):
    """
    Harmonises New Zealand tenure categories into the broad
    tenure categories used for comparison with Australian G37 data.

    Output categories:
    - Owned outright
    - Owned with mortgage
    - Rented
    """

    df = df.copy()

    # The first column contains the NZ tenure category
    category_col = df.columns[0]

    # The second column contains the household count
    value_col = df.columns[1]

    # Create a new column for the harmonised tenure category
    df["Tenure Type"] = None

    # OWNED OUTRIGHT
    # Includes:
    # - directly owned/partly owned dwellings with no mortgage payments
    # - family trust dwellings with no mortgage payments

    owned_outright = (
        df[category_col].str.contains("dwelling owned or partly owned, mortgage payments not made", case=False, na=False)
        |
        df[category_col].str.contains("dwelling held in a family trust, mortgage payments not made", case=False, na=False)
    )

    df.loc[owned_outright, "Tenure Type"] = "Owned outright"

    # OWNED WITH MORTGAGE
    # Includes:
    # - directly owned/partly owned dwellings with mortgage payments
    # - family trust dwellings with mortgage payments

    owned_with_mortgage = (
        df[category_col].str.contains("dwelling owned or partly owned, mortgage payments made", case=False, na=False)
        |
        df[category_col].str.contains("dwelling held in a family trust, mortgage payments made", case=False, na=False)
    )

    df.loc[owned_with_mortgage, "Tenure Type"] = "Owned with mortgage"

    # RENTED
    # Includes households that do not own or hold the dwelling
    # in a family trust and make rent payments

    rented = df[category_col].str.contains("dwelling not owned and not held in a family trust, rent payments made", case=False, na=False)

    df.loc[rented, "Tenure Type"] = "Rented"

    # Remove categories that do not have a clear equivalent in the Australian comparison
    df = df.dropna(subset=["Tenure Type"]).copy()

    # Ensure household counts are numeric
    df[value_col] = pd.to_numeric(df[value_col], errors="coerce")

    # Combine NZ categories that belong to the same harmonised tenure category
    df = df.groupby("Tenure Type", as_index=False)[value_col].sum()

    # Standardise the household count column name
    df = df.rename(columns={value_col: "Household Count"})

    return df
```

```python
# we only need tenures we can compare between AU and NZ
comparable_tenures = ["Owned outright", "Owned with mortgage", "Rented"]
au_q1_comparable = clean_g37[clean_g37["Tenure Type"].isin(comparable_tenures)].copy()

# harmonise nz for q1
harmonised_nzq1 = harmonise_nz_tenure(clean_nzq1)
```

```python
def calculate_proportions(df):
    """
    Calculates the percentage share of each tenure category
    within the comparable housing tenure categories.
    """

    df = df.copy()

    # Total households across the comparable tenure categories
    total_households = df["Household Count"].sum()

    # Calculate percentage for each tenure type
    df["Proportion (%)"] = (df["Household Count"] / total_households) * 100

    return df
```

```python
# obtain the relevant proportions
au_q1_proportion = calculate_proportions(au_q1_comparable)
nz_q1_proportion = calculate_proportions(harmonised_nzq1)
```

**Limitation**

A key limitation of this comparison is that the Australian and New Zealand Census tables use different structures and levels of detail for housing tenure. The Australian G37 table already groups households into broad categories such as Owned outright, Owned with mortgage, and Rented, whereas the New Zealand table separates tenure into more detailed categories based on ownership arrangement and payment status, including whether a dwelling is directly owned, held in a family trust, and whether mortgage or rent payments are made. #### **Sub-Question 2 — Income and Affordability**
**Among renting households, how do rent and income levels and their distributions compare between Australia and New Zealand, and what does this reveal about housing affordability pressures in each country?**

- Tables Used AU: `clean_g40`, `clean_g33`
- Tables Used NZ: `clean_household_income_rent`, `clean_rent_dist`, `clean_income_dist`

Affordability is visualised as a rent-to-income ratio **filled bar chart** comparing AU and NZ directly, alongside separate rent and income distribution bar charts (shown side by side by country) to reveal the difference in affordability pressures between the countries and whether it predominantly stems from high rent costs or low income

**Filled Rent-Income Ratio Bar Charts**

Our first visual is a **filled bar chart** where each bar is split into the share of income spent on rent and the remaining income, benchmarked against a 30% housing stress threshold [13]

Since census data reports household counts within brackets rather than exact dollar values, average weekly rent and income are estimated using each bracket's midpoint as a representative value and then weighted by its corresponding household count: `Σ(midpoint × household count) / Σ(household count)`.

This method is applied consistently across both countries' rent and income tables to give a reasonable average income and rent for both countries.

 Open-ended top brackets (e.g. "$950 or more") have no natural midpoint, so a representative value is assumed for each, outlined below

| Bracket (AUD/week) | Country | Share of households | Assumed value (AUD/week) | Why |
|---|---|---|---|---|
| Rent: 950 or more | AU | 1.79% | 1,050 | A small share of households, so the exact value has little effect on the result. Set to the previous bracket's upper bound (950) plus the same ~100 width used throughout the table. |
| Income: 4,000 or more | AU | 13.61% | 6,000 (1.5× threshold) | A large enough share that a low guess would noticeably skew the result. High-income brackets tend to have long tails, so a value 1.5× the bracket's starting point is used as a representative estimate. |
| Rent: 736 and over | NZ | 6.46% | 836 | Same reasoning as AU's rent bracket — a comparatively thin tail, so set using the same "previous bracket plus its width" approach. |
| Income: 3,538 or more | NZ | 8.84% | 5,307 (1.5× threshold) | Same underlying issue and same 1.5× rule as AU's income bracket, applied consistently across both countries. |

**Note:** Due to the assumptions made for open-ended brackets, the exact dollar figures in this analysis should be treated as approximations rather than precise values. However, since the same estimation approach is applied consistently to both countries, the *relative* comparison between countries remains reliable even if the absolute numbers carry some uncertainty.

```python
# Assumed representative values for open-ended top brackets from assumptions table above
TOP_RENT_AU = 1050      # AU rent: "$950 or more"
TOP_INCOME_AU = 6000    # AU income: "$4,000 or more" (1.5 x threshold)
TOP_RENT_NZ = 836       # NZ rent: "$736 and over"
TOP_INCOME_NZ = 5307    # NZ income: "$3,538 or more" (1.5 x threshold)

def midpoint_from_bracket(x, top_value):
    """
    Return the midpoint of a bracket label such as '$350-449'.
    Handles lower-end ranges (under/less), middle ranges (A-B) and
    upper-end ranges (A or more / A and over). For open-ended top
    brackets there is no observable midpoint, so top_value is returned.
    """
    x = x.replace('$', '').replace(',', '')  # remove $ signs and thousands separators
    if 'over' in x or 'more' in x:
        return top_value  # open-ended top bracket, based on stated assumptions
    if '-' in x:
        lo, hi = x.split('-')
        return (int(lo.strip()) + int(hi.strip())) / 2
    if 'less' in x or 'Under' in x:
        upper = int(re.search(r'\d+', x).group())  # first sequence of digits in the label
        return upper / 2  # assume the bracket runs from $0 to the stated value
    raise ValueError(f"Unrecognised bracket: {x}")  # fail loudly rather than return None

def add_midpoints(df, bracket_col, new_col, top_value):
    """Add a column of bracket midpoints to df, using top_value for the open-ended bracket."""
    df[new_col] = df[bracket_col].apply(lambda b: midpoint_from_bracket(b, top_value))

def weighted_mean(values, weights):
    """Average of values weighted by household counts: sum(value * count) / sum(count)."""
    return (values * weights).sum() / weights.sum()

# Midpoints for AU and NZ (top values come from the constants above)
add_midpoints(clean_g40, 'Rent Bracket', 'Rent Midpoint', TOP_RENT_AU)
add_midpoints(clean_g33, 'Income Bracket', 'Income Midpoint', TOP_INCOME_AU)
add_midpoints(clean_household_income_rent, 'Total Household Income', 'Income Midpoint', TOP_INCOME_NZ)
add_midpoints(clean_household_income_rent, 'Weekly Rent Paid By Household', 'Rent Midpoint', TOP_RENT_NZ)

# Weighted average weekly rent and income for AU
weighted_rent_avg_au = weighted_mean(clean_g40['Rent Midpoint'], clean_g40['Number Of Households'])
weighted_income_avg_au = weighted_mean(clean_g33['Income Midpoint'], clean_g33['Household Count'])

# Weighted average weekly rent and income for NZ
# Both come from the same joint income-by-rent table, so they share the same household counts as weights
weighted_rent_avg_nz = weighted_mean(clean_household_income_rent['Rent Midpoint'], clean_household_income_rent['Household Count'])
weighted_income_avg_nz = weighted_mean(clean_household_income_rent['Income Midpoint'], clean_household_income_rent['Household Count'])

# Rent-to-income ratio (proportion of income spent on rent) and the remaining proportion
au_ratio = weighted_rent_avg_au / weighted_income_avg_au
au_remainder = 1 - au_ratio

nz_ratio = weighted_rent_avg_nz / weighted_income_avg_nz
nz_remainder = 1 - nz_ratio


def proportional_filledchart(countries, proportion, data_values, colours, threshold, title):
    """
    Plots a stacked bar chart of proportion vs remainder per country, with a benchmark line.
    """
    fig, ax = plt.subplots(1, figsize=(7, 5))
    # Reshape the flat lists into a country x proportion matrix, then draw as a stacked bar chart
    # iterates over countries, proportion, data_values by index and then assigns value to the (country, proportion) pair in the matrix
    pd.crosstab(index=countries, columns=proportion, values=data_values, aggfunc='sum').plot(
    kind='bar', stacked=True, ax=ax, color=colours, edgecolor='white', linewidth=0.5) # plot a stacked bar chart which has already been normalised

    # Draw the benchmark line at the given threshold, with a dynamic label matching its value
    ax.axhline(threshold, color='#B85C5C', linestyle='--', linewidth=1.5, alpha=0.8, label=f'{round(threshold * 100)}% threshold')
    ax.set_xlabel('Country', fontsize=12)
    ax.set_ylabel('Proportion', fontsize=12)
    ax.set_title(title, fontsize=15, weight='bold', pad=12)
    ax.tick_params(axis='x', rotation=0, labelsize=10)
    ax.tick_params(axis='y', labelsize=10)
    ax.grid(axis='y', alpha=0.3); ax.grid(axis='x', visible=False)  # keep horizontal gridlines only, for readability against the stacked bars
    sns.despine(ax=ax, left=False, bottom=False)
    ax.legend(loc='upper center', bbox_to_anchor=(0.5, -0.12), ncol=3, frameon=False) # change legend to a column-based figure 
    plt.tight_layout()
    plt.show()

# Plot AU vs NZ's rent-to-income proportion, each split into "to rent" and "remaining" segments, against the 30% housing stress benchmark
proportional_filledchart(['Australia', 'New Zealand', 'Australia', 'New Zealand'],
                          ['Proportion to Rent', 'Proportion to Rent', 'Remaining Proportion', 'Remaining Proportion'],
                          [au_ratio, nz_ratio, au_remainder, nz_remainder], [AU_COLOUR, '#E8E1D5'], 0.3, 'Proportion of Income Contributing to Rent by Country')
```

**Limitation - Income population scope**: AU's income figure (G33) covers all households (owners and renters combined), since no AU table reports income specifically for renters. NZ's income figure is renter-specific and as renting households typically earn less than owners on average, AU's income denominator is likely inflated relative to true renter income — meaning AU's calculated rent-to-income ratio is probably understated relative to renters' actual affordability pressure.

**Limitation - Currency conversion and census-year mismatch**: `convert_bracket_to_aud` uses the 2023 average NZD/AUD exchange rate, matching New Zealand's census year. #### **Sub-Question 3 - Affordability in regards to Housing-Cost Poolings**

We visualise housing affordability through comparing the proportion of family and group households between Australia and New Zealand, as well as the size composition of group households in both countries. Group housing — people pooling housing costs by sharing a dwelling — is treated as a proxy measure of housing affordability, where higher rates and larger group sizes indicate poorer affordability.

A household is classified as Family if it contains at least one family unit (related people, per the ABS/Stats NZ definition), even where unrelated individuals also reside there; all other multi-person households are classified as Group. Single-resident households are grouped with Family households in both countries, since living alone does not represent housing-cost pooling.

First we merge both sets of tables to allow for easier side by side comparison:

```python
# function to add a proportions column to df based on a select column
def proportions(df, value_col_name):
    """
    Add a 'proportion' column (share of the total of value_col_name)
    to the DataFrame passed in, and return it.
    Values are left unrounded so charts and text can format them as needed.
    """
    total = df[value_col_name].sum()
    df["proportion"] = df[value_col_name] / total
    return df

# add proportions column to tables
nz_family_nonfamily = proportions(nz_family_nonfamily, "Number Of Households")
nz_nonfamily_composition = proportions(nz_nonfamily_composition, "Number Of Households")
au_family_nonfamily = proportions(au_family_nonfamily, "Number Of Households")
au_nonfamily_composition = proportions(au_nonfamily_composition, "Number Of Households")

# add country column to tables
nz_family_nonfamily["Country"] = "NZ"
nz_nonfamily_composition["Country"] = "NZ"
au_family_nonfamily["Country"] = "AU"
au_nonfamily_composition["Country"] = "AU"

# concat matching table pairs
nonfamily_composition_comparison = pd.concat([au_nonfamily_composition, nz_nonfamily_composition], ignore_index=True)
family_nonfamily_comparison = pd.concat([au_family_nonfamily, nz_family_nonfamily], ignore_index=True)
```

**Assumptions:**
- Group households are assumed to reflect housing-cost pooling rather than cultural or social living preferences, for the purposes of this affordability comparison.
- If a household is categorised as a family household, it is not an example of housing-cost pooling even if there are also non-related residents living there

**Limitations:**
- Due to AU database structure, NZ's data had to be simplified down so that they both have the same Family and Non-family categories.
- Group housing resident counts in the AU database only go up to "6 or more," while New Zealand's goes up to "8 or more." NZ's 6+ categories were aggregated into a single "6+" bracket to match AU, reducing NZ's data specificity.
- Approximately 4% of NZ households fall under "composition unidentifiable" and are excluded from both totals, since they cannot be classified as Family or Group.
- Due to the family household definition, we can't identify cases of house-cost pooling in family homes (if an unrelated resident also lives there)

This suggests that a larger segment of New Zealand's population — particularly those in larger shared households — may be experiencing housing affordability pressure severe enough to require multi-person cost-sharing, a form of housing insecurity less pronounced among Australia's predominantly two-person group households.

#### **Sub-Question-4 - Bedrooms vs Residents**

**How does housing adequacy — approximated through the difference in bedroom count and household size — vary across household compositions in Australia (2021) and New Zealand (2023), and what does this reveal about which household types face the greatest risk of overcrowding in each country?**

- Tables Used AU: `clean_g29`, `clean_g35`, `clean_g41`, `clean_g42`
- Tables Used NZ: `clean_nzq4`


The difference in bedroom count and household size is visualised using a **diverging bar chart** comparing AU and NZ directly on the same visual for each household composition type. Bars extend right of zero when average bedrooms exceed the household size (a surplus - indicating a lower risk of overcrowding) while bars extend left of zero when household size exceeds the average number of bedrooms (a deficit - indicating higher risk of overcrowding). Analysing the overcrowding risk of the 2 countries side-by-side through this bedroom-to-household-size difference reveals which household types in each country struggling to find housing that accommodates their needs, contributing to the overall picture of housing security across both countries.
#
**Bedroom-Household Size Diverging Bar Charts**

Since census data reports household counts within brackets rather than a single bedroom count or resident count per household, average bedrooms and average household size are each estimated using the same method as in Q2: weighted by household count across all rows for a given composition type - `Σ(value × household count) / Σ(household count)`.

For New Zealand, this is computed directly from the single joint table linking bedrooms, composition, and resident count. For Australia, no such joint table exists, so average bedrooms per composition type is instead built in two steps: a weighted average bedroom count per dwelling structure (G41), combined with each composition type's distribution across dwelling structures (G42) - while average household size is computed separately from G29 (family types) and G35 (group/lone-person households).

Open-ended top brackets (e.g. "6 or more bedrooms") have no natural value, so a representative value is assumed. We have decided to apply a method of +1 the top boundary as a rough estimate for all top bracket estimations to maintain consistency. Unlike Q2's income brackets, bedroom and resident counts do not have long right tails since very few households have very large numbers, so a simple +1 on the top boundary was deemed suitable.

```python
# NZ Room-Household Size Surplus/Deficit Table
bedroom_cols = ['One bedroom', 'Two bedrooms', 'Three bedrooms', 'Four bedrooms', 'Five or more bedrooms']
bedroom_values = [1, 2, 3, 4, 6]  # numeric bedroom count per column; "five or more" assumed as 6

nz_q4_df = clean_nzq4.copy()
nz_q4_df['Total Households'] = nz_q4_df[bedroom_cols].sum(axis=1)  # households across all bedroom brackets, per row
nz_q4_df['Bedroom Weighted Sum'] = sum(nz_q4_df[col] * val for col, val in zip(bedroom_cols, bedroom_values))  # Σ(bedrooms × count) per row (applied on series' so across the whole df)
nz_q4_df['People Weighted Sum'] = nz_q4_df['Number Of People'] * nz_q4_df['Total Households']  # people × household count per row

# Aggregate to one row per composition type
nz_q4_result = nz_q4_df.groupby('Composition Type')[['Bedroom Weighted Sum', 'People Weighted Sum', 'Total Households']].sum().reset_index() # sum per composition type

# Compute weighted averages by dividing by the total number of households for the corresponding composition Type
nz_q4_result['Avg Bedrooms'] = nz_q4_result['Bedroom Weighted Sum'] / nz_q4_result['Total Households']
nz_q4_result['Avg People'] = nz_q4_result['People Weighted Sum'] / nz_q4_result['Total Households']

# Calculate surplus by subtracting avg number of bedrooms by avg number of people
nz_q4_result['Surplus'] = nz_q4_result['Avg Bedrooms'] - nz_q4_result['Avg People']
nz_q4_result = nz_q4_result.sort_values(by='Surplus', ascending=False)[['Composition Type', 'Avg Bedrooms', 'Avg People', 'Surplus']] # sort results and select only required columns
nz_q4_result
```

Our goal now is to replicate the exact table above for AU

**Limitation — Bedroom availability assumed independent of household composition within a dwelling structure**

Step 2 assumes that, within a given dwelling structure type, bedroom count doesn't vary systematically by which household composition occupies it. Instead households are treated as if they self-select into a dwelling structure but not into a specific unit size within it. In reality this likely understates true variation, since households probably do self-select into unit size (e.g. families will have larger apartments whereas lone households will have smaller apartments)

However, this risk is not uniform across dwelling structures. It is likely less significant for dwelling structures with a narrow, fairly fixed bedroom range such as flats and apartments which are typically 1–3 bedrooms regardless of occupant. This leaves little room for bedroom count to meaningfully differ by household composition in the first place. The assumption carries more risk for structures like separate houses, where bedroom counts vary more widely (3 up to 6+), meaning genuine differences between, for example, a couple's house and a family of 4's house are likely to be averaged away. That said, this risk is likely mitigated by affordability given few households of any composition, including larger families, can afford homes at the top of the bedroom range, so most houses likely cluster closer to the middle.

**Limitation — Family size used as household size**

For family household types, G29's family counts and persons-in-family stand in for household counts and household size. G29 counts only the people in the family unit (couples without children average exactly 2.0), so any unrelated residents or second families sharing the dwelling are missed. This likely slightly understates Australian household size, and so slightly overstates the bedroom surplus.

```python
# Step 1 - Calculate the Average number of Rooms per Dwelling Structure
def weighted_avg_by_group(data, group_col, value_col, weight_col):
    """
    Count-weighted average of value_col within each group: Σ(value × weight) / Σ(weight).
    Here each row is a bracket's representative value (e.g. bedroom count) and
    weight_col is its household count. Returns [group_col, 'Avg <value_col>'].
    """
    data = data.copy() # create a copy so we don't alter the original dataframe
    data['weighted'] = data[value_col] * data[weight_col] # create a new column containing the weighted multiplication
    grouped = data.groupby(group_col)[['weighted', weight_col]].sum() # group by and aggregate on weighted column and the weights
    result = (grouped['weighted'] / grouped[weight_col]).reset_index(name=f'Avg {value_col}') # perform a division between weighted values and weights to get weighted average
    return result

avg_bedrooms_per_dwelling = weighted_avg_by_group(clean_g41, 'Dwelling Type', 'Bedroom Count', 'Household Count')
avg_bedrooms_per_dwelling

# Step 2 - Merge the avg_bedrooms_per_dwelling table with g42 so we can visualise the avg bedroom count of each composition type by dwelling type
avg_bedroom_join_composition = avg_bedrooms_per_dwelling.merge(clean_g42)
avg_bedroom_join_composition

# Step 3 - Now we want to calculate the weighted average bedroom count per composition type (We used Dwelling Type to link the 2 tables by an estimate)
avg_bedrooms_per_composition = weighted_avg_by_group(avg_bedroom_join_composition, 'Composition Type', 'Avg Bedroom Count', 'Household Count') # apply weighted avg calculation function
avg_bedrooms_per_composition = avg_bedrooms_per_composition.rename(columns={'Avg Avg Bedroom Count': 'Avg Bedrooms'}) # rename to required name
avg_bedrooms_per_composition

# Drop Other Family — no NZ equivalent category exists for comparison
avg_bedrooms_per_composition = avg_bedrooms_per_composition[avg_bedrooms_per_composition['Composition Type'] != 'Other Family'].reset_index(drop=True)

# Step 4 - Compute Average Household Sizes for Family Households and Group Households
clean_g29['Avg People'] = clean_g29['Number Of People'] / clean_g29['Number Of Households'] # Avg no. People in the household for families
clean_g35['Avg People'] = clean_g35['Number Of People'] / clean_g35['Number Of Households'] # Avg no. People in the household for non family households

# We can then concatenate them on top of each other to form a single df of Average People per Composition
avg_people_per_composition = pd.concat([clean_g29[['Composition Type', 'Avg People']], clean_g35[['Composition Type', 'Avg People']]], ignore_index=True)

# Step 5 - Merge Average Household Sizes Per Composition table with Average Bedrooms Per Composition Table
# We can finally merge with avg_bedrooms_per_composition so we now have a table with Avg People and Avg Bedrooms - exactly like NZ
au_q4_result = avg_bedrooms_per_composition.merge(avg_people_per_composition)
au_q4_result['Surplus'] = au_q4_result['Avg Bedrooms'] - au_q4_result['Avg People']
au_q4_result = au_q4_result.sort_values(by='Surplus', ascending=False)
```

```python
# The idea is to now concatenate AU and NZ tables and then plot as a side-by-side diverging bar chart
def plot_q4_surplus(au_result, nz_result):
    """
    Diverging bar chart of bedroom surplus (avg bedrooms − avg household size)
    by composition type, AU vs NZ. Positive = surplus, negative = deficit.
    """
    au_result = au_result.copy()
    nz_result = nz_result.copy()

    # add the country label to both dataframes and concatenate them into 1 dataframe
    au_result['Country'] = 'Australia'
    nz_result['Country'] = 'New Zealand'
    data = pd.concat([nz_result, au_result])

    plt.figure(figsize=(11, 5))

    # plot the diverging bar chart (horizontal bars)
    fig = sns.barplot(data=data, x='Surplus', y='Composition Type', hue='Country', # hue of country allows us to plot bars for each country
        palette={'Australia': AU_COLOUR, 'New Zealand': NZ_COLOUR}, errorbar=None)

    # add the surplus value (rounded to 2dp) for each bar
    for container in fig.containers:
        labels = [f'{bar.get_width():+.2f}' for bar in container]
        fig.bar_label(container, labels=labels, padding=3)

    sns.despine(left=False, bottom=False)
    plt.title('Bedroom Surplus by Household Composition: Australia and New Zealand', fontsize=15, weight="bold", pad=12)
    plt.xlabel('Bedrooms minus residents (surplus +, deficit −)')
    plt.grid(False)
    plt.tight_layout()
    plt.show()

plot_q4_surplus(au_q4_result, nz_q4_result)
```

**Interpretation and Analysis**

Comparing the raw surplus figures reveals that Australia has a significantly larger bedroom surplus than New Zealand for smaller households, specifically - Lone Person: 1.89 vs 1.44 and Couple, No Children: 1.15 vs 1.08. This suggests that Australian households without children have comparatively more rooms than residents and as such a likely lower risk of overcrowding. Conversely, Australia shows a larger deficit for Couple with Children at -0.56 vs -0.32, which indicates that Australian households with children face significantly greater overcrowding risk compared to New Zealand.

Despite Australia having more composition types with greater surpluses, this matters disproportionately for overall housing security since family households make up the majority of both countries' households at roughly 71% as cited by the AIFS [2] and Stats NZ [3]. A larger surplus concentrated in smaller, less common household types does less to improve the overall national housing security than a larger deficit concentrated in the most common household types.

Therefore, according to this analysis, Australia's overall housing security is likely to be **worse** than New Zealand's from the perspective of household overcrowding, despite showing better raw numbers for the majority of composition types, since the main category where Australia underperforms, couples with children, is a type of family household, and family households account for 71% of households in both countries. Additionally, we can conclude that couples with children are the most susceptible to housing insecurity from the perspective of space adequacy, being the only household type with a deficit of bedrooms, while all other households are reasonably secure.

However, this comparison should be read with some caution as Australia's estimates are built through a chained approximation with multiple steps (unlike New Zealand's direct figures), which likely compresses the true range of outcomes. Because the method assumes bedroom counts do not vary by household type within a dwelling type, estimates are pulled towards the overall average, so the largest errors fall on the household types at either end. This likely overstates Australia's true surplus for smaller households, and overstates the true severity of its deficit for family households with children. Accounting for this, the Couple with Children deficit appears in both countries independently, so the finding that this group faces the greatest space pressure is more reliable than the exact size of the gap between the two countries.

## **Practical Insights and Future Action**

In Australia, couples with children may face the greatest overcrowding risk because their estimated bedroom availability falls 0.56 bedrooms below their average household size—the largest shortfall among the household types analysed. If this reflects overcrowded conditions, the AIHW reports that limited space can disrupt children’s study, play and sleep and is associated with poorer school performance, reduced privacy, family conflict and mental health problems [5]. Australian housing agencies should therefore use household-level occupancy data to identify affected families and determine whether appropriately sized homes are available and affordable. Commonwealth Rent Assistance was already operating during the 2021 Census [6], while the Housing Australia Future Fund was established in 2023 [7]. These measures could be improved by helping low-income families access homes with sufficient bedrooms and ensuring new housing programs provide an appropriate mix of affordable, family-sized dwellings.

In New Zealand, renters and larger group households may experience greater housing affordability pressure. Estimated average weekly rent equals approximately 23% of estimated average renter income, compared with approximately 17% of all-household income in Australia (21% using Australia's renter-income sensitivity estimate). This higher burden may leave households with less income for essential expenses, savings and unexpected costs. New Zealand's higher proportion of group households with three or more residents is also consistent with greater reliance on multi-person cost-sharing, which may limit housing choice and reduce privacy when dwellings are unsuitable. The government provides the Accommodation Supplement [8] and was delivering additional public and transitional housing through the Public Housing Plan 2021–2024 at the time of the 2023 Census [9]. However, the 2022 New Zealand Income Support Survey found low Supplement uptake among potentially eligible people not receiving a main benefit [10]. Government agencies should improve awareness and access, assess whether payment rates reflect local rents, provide suitable affordable housing in high-pressure areas, and investigate whether larger shared households reflect personal preference or financial necessity.

## **Conclusion**

We have examined the housing security in Australia (2021) and New Zealand (2023) across 3 main aspects: tenure (Sub-Question 1), affordability (Sub-Questions 2 and 3) and adequacy (Sub-Question 4). Examining tenure distribution showed that Australia had higher outright ownership indicating better long-term housing stability than New Zealand. The affordability measures were consistent: Australia showed a lower rent-to-income ratio and less housing-cost pooling than New Zealand, suggesting New Zealand's renters and shared-housing residents face greater affordability pressure than their Australian counterparts. Adequacy pointed the other way — Australia's larger bedroom deficit for couples with children, a major family household type in both countries, gives it a worse overcrowding risk than New Zealand when the size of each household type is taken into account, despite better raw figures for less common household types.

Housing security therefore cannot be reduced to a single verdict, nor to a single population: based on the data analysed, Australia's renters and cost-pooling households are comparatively better off than New Zealand's, while Australia's families with children likely face greater overcrowding risk than New Zealand's. This distinction would have been missed by a single-dimension comparison, and underscores that addressing either country's housing challenges requires targeting genuinely different populations under genuinely different pressures.
